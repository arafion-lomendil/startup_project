## Шаг 2. Инжиниринг признаков

### 2.1. Группы по срокам финансирования
Разделим все компании на три группы:

* Единичное финансирование — был всего один раунд финансирования.

* Срок финансирования до года — между первым и последним раундом финансирования прошло не более года.

* Срок финансирования более года.

Визуализируем соотношение этих групп, создав два графика:

* По количеству компаний: покажем, какой процент от общего числа компаний относится к каждой из трёх групп.
* По объёму инвестиций: отобразим, какую долю от общего объёма привлечённых средств получила каждая группа.

```python
# Создаем функцию для классификации компаний
def finance_groups(row):
    if row['funding_rounds'] == 1:
        return 'Единичное финансирование'
    else:
        if (row['last_funding_at'] - row['first_funding_at']).days <= 365:
            return 'Срок финансирования до года'
        else:
            return 'Срок финансирования более года'

# Применяем функцию и создаем столбец с новыми значениями
df['funding_groups'] = df.apply(finance_groups, axis=1)
# Группируем показатели по количеству компаний и суммарному объему инвестиций
company_count = df.groupby('funding_groups')['name'].count()
investment_sum = df.groupby('funding_groups')['funding_total_usd'].sum()
```

Теперь визуализируем полученные группировки на круговых диаграммах

```python
# Задаем размер графика
plt.figure(figsize=(10,5))

# Рисуем круговую диаграмму
company_count.plot(
    kind='pie',    
    autopct='%1.1f%%',
    startangle=90
)

# Настраиваем оформление диаграммы
plt.title('Соотношение групп финансирования по количеству компаний\n')
plt.ylabel('')
plt.axis('equal')

# Выводим график
plt.show()
```

Большая часть компаний (**59%**) относится к компаниям с единичным финансированием, тогда как оставшиеся **41%** разделяют компании с финансированием до года (**11%**) и более года (**30%**).

```python
# Задаем размер графика
plt.figure(figsize=(10,5))

# Рисуем круговую диаграмму
investment_sum.plot(
    kind='pie',    
    autopct='%1.1f%%',
    startangle=90
)

# Настраиваем оформление диаграммы
plt.title('Соотношение групп финансирования по суммарному объему инвестиций\n')
plt.ylabel('')
plt.axis('equal')

# Выводим график
plt.show()
```

По суммарному объему инвестиций компании, которые финансировали более года ожидаемо лидируют (**61,9%**). Однако компании с финансированием до года сильно уступают компаниям с единичным финансированием в объеме инвестиций (**7.5%** против **30.6%**).


### 2.2 Выделение средних и нишевых сегментов рынка

Компании указывают свой сегмент рынка в столбце `market`. Рассчитаем, как часто в датасете встречается каждый из сегментов. Сегменты, к которым относится более 120 компаний, отнесем к массовым, сегменты, в которые входит от 35 до 120 включительно будем считать средними, а сегменты до 35 компаний будут нишевыми. Рассчитаем, сколько сегментов попадает в каждую из категорий. Также построим график распределения количества компаний в сегментах и отобразим на нём разделение на нишевые и средние сегменты.

```python
# Группируем количество компаний для каждого рынка
market_count = df.groupby('market')['name'].transform('count')

# Определяем функцию категоризации рынка
def market_category(value):
    if pd.isna(value):
        return None
    if value > 120:
        return 'mass'
    elif value >= 35:
        return 'mid'
    else:
        return 'niche'

# Применяем функцию к новому столбцу market_cat с категориями mass, mid и niche
df['market_cat'] = market_count.apply(market_category)
# Рассмотрим какое количество сегментов и компаний в сегмента присутствует в каждой категории рынка
mass_mid_niche = df.groupby('market_cat').agg({
    'market': 'nunique',
    'name': 'count'
})
mass_mid_niche
```

```python
# Формируем сводную таблицу с группировкой по категориям и сегментам
pivot_mass_mid_niche = pd.pivot_table(
    data=df,
    index=['market_cat','market'],
    values='name',
    aggfunc='count'
)
# Выводим первые 5 позиций какждой категории
pivot_mass_mid_niche.groupby('market_cat').head()
```

Для создания графика с разделением на нишевые и средние сегменты необходимо исключить массовые сегменты, для чего создадим еще одну функцию.

```python
# Определение функции для niche и mid сегментов
def market_mid_niche(value):
    if value >= 35 and value <= 120:
        return 'mid'
    if value < 35:
        return 'niche'
    else:
      return None

# Применяем функцию к новому столбцу market_mid_niche
df['market_mid_niche'] = market_count.apply(market_mid_niche)
mid_niche = df.groupby('market_mid_niche')['name'].count()
mid_niche
```

```python
# Задаем размер графика
plt.figure(figsize=(10,5))

# Рисуем круговую диаграмму
mid_niche.plot(
    kind= 'pie',
    autopct='%1.1f%%',
    startangle=90,
    labels=['niche', 'mid']
)
# Настраиваем оформление диаграммы
plt.title('Распределение рынков по средним и нишевым сегментам')
plt.axis('equal')
plt.show()
```

Оставим в столбце `market` только массовые сегменты. Для остальных сегментов заменим значения на заглушки — `niche` для нишевых и `mid` для средних.

Дальнейшие исследования будем выполнять с учётом этой замены. Индивидуальные сегменты внутри средней и нишевой групп рассматривать не нужно — они объединяются в два общих сегмента.


```python
# Заменяем нишевые и средние сегменты на заглушки
df.loc[df['market_cat'] == 'mid', 'market'] = 'mid'
df.loc[df['market_cat'] == 'niche', 'market'] = 'niche'
```

По разделению сегментов компаний на массовые, средние и нишевые можно отметить следующее:
- К массовым сегментам относится подавляющее большинство компаний, однако наибольшее количество уникальных сегментов относится к нишевым;
- На круговой диаграмме видно, что и в сравнении с компаниями со средним сегментом, нишевых гораздо больше (**82.2%** против **17.8%**).

Судя по 25 и 75 процентилям чаще всего компании финансируют в пределах от 350000 до 10000000 долларов США.


Теперь определим компании с аномальным объемом общего финансирования с использованием метода IQR отдельно по каждому сегменту. Определим сегменты рынка с наибольшей долей компаний, получивших аномальное финансирование, и выведем топ таких сегментов.

```python
# Для типичной выборки данных вычисляем квантили и IQR
Q1 = df['funding_total_usd'].quantile(0.25)
Q3 = df['funding_total_usd'].quantile(0.75)
IQR = Q3 - Q1
# Определяем диапазон iqr для каждой категории сегментов
typical_values = df.groupby('market')['funding_total_usd'].agg(
  lambda x: IQR
)
# Вывадим типичное среднее значение всех категорий рынка
print(f'Типичные значения лежат в диапазоне: ${typical_values.mean().round(2)}')
```

```python
# Определияем сегменты рынка с наибольшей долей компаний, получивших аномальное финансирование
iqr_stats = df.groupby('market')['funding_total_usd'].agg([
    ('IQR', lambda x: IQR),
    ('Q3', lambda x: Q3),
     ('max_funding', 'max'),
     ('total_funding', 'sum'),
  ('abnormal_funding', lambda x: x.max() - Q3),
    ('abnormal_count', lambda x: (x > Q3).sum())
     ]).round(2)
# Вычисляем суммарное финансирование по всем компаниям
overall_funding = df['funding_total_usd'].sum()
# Выявляем процент аномального финансирования по сегментам
iqr_stats['share_of_total'] = (iqr_stats['abnormal_funding'] / overall_funding * 100).round(2)
# Выводим топ-5 сегментов с максимально аномальной долей финансирования
iqr_stats.sort_values(by='share_of_total', ascending = False).head()
```

Наибольшую долю аномального финансирования занимает сегмент `mobile` - 4,62% от всего финансирования. Менее процента составляют доли `finance`, `internet`, `health and wellness` и *средней категории сегментов*. <span style="color: blue"> По количеству самих компаний, лидером в этом топе являются компании среднего сегмента и только потом сегменты `mobile` и `finance`.</span>

### 4.1 Динамика предоставления финансирования по годам

Используя столбцы `funding_total_usd` и `funding_rounds`, рассчитайте для каждой компании средний объём одного раунда финансирования.

На основе получившейся таблицы постройте графики, отражающие:
* динамику типичного размера средств, которые стартапы получали в рамках одного раунда финансирования;

* динамику общего количества раундов за каждый год, то есть насколько активно происходили инвестиции на рынке (чем больше раундов, тем выше активность).

Когда будете строить графики в этом задании и следующих, используйте данные только по тем компаниям, которые остались в датасете после предыдущих фильтраций.

На основе полученных данных ответьте на вопросы:
* В каком году типичный размер средств, собранных в рамках одного раунда, был максимальным?

* Какая тенденция наблюдалась в 2014 году по количеству раундов и средств, выделяемых в рамках каждого раунда?

```python
# Сначала создаем столбец с размером финансирования за раунд
df['funding_per_round'] = df['funding_total_usd'] / df['funding_rounds']
# Теперь создаем сводную таблицу с новым показателем
pivot_total_usd_rounds = pd.pivot_table(
    data=df,
    index=['name'],           
    values=['funding_total_usd', 'funding_rounds','funding_per_round'],
    aggfunc={'funding_per_round': 'mean', 'funding_total_usd': 'sum', 'funding_rounds': 'sum'}
)
pivot_total_usd_rounds = pivot_total_usd_rounds[['funding_total_usd', 'funding_rounds', 'funding_per_round']]
pivot_total_usd_rounds
```

```python
# Срздаем новый столбец с годом на основании даты последнего финансирования компании
df['year'] = pd.to_datetime(df['last_funding_at']).dt.year
# Группируем средний размер финансирования за раунд по годам
funding_per_year = df.groupby('year')['funding_per_round'].mean().reset_index()
funding_per_year
```

```python
# Теперь группируем количество раундов финансирования за каждый год
round_per_year = df.groupby('year')['funding_rounds'].sum().reset_index()
round_per_year
```

```python
# Создаем фигуру с заданным размером
fig, ax1 = plt.subplots(figsize=(10, 6))

#  Первый график – размер средств
ax1.plot(
    funding_per_year['year'],
    funding_per_year['funding_per_round'],
    marker='o',
    linewidth=2,
    markersize=5,
    color='tab:blue',
    label='Средний размер финансирования (млн $)'
)
ax1.set_xlabel('Год')
ax1.set_ylabel('Размер средств (млн $)', color='tab:blue')
ax1.tick_params(axis='y', labelcolor='tab:blue')
ax1.grid(alpha=0.5)

# Второй график - количество раундов
ax2 = ax1.twinx()   
ax2.plot(
    round_per_year['year'],
    round_per_year['funding_rounds'],
    marker='o',
    linewidth=2,
    markersize=5,
    color='tab:red',
    label='Количество раундов'
)
ax2.set_ylabel('Количество раундов', color='tab:red')
ax2.tick_params(axis='y', labelcolor='tab:red')

plt.title('Динамика финансирования по годам', fontsize=14)
plt.xticks(funding_per_year['year'], rotation=45)
lines_1, labels_1 = ax1.get_legend_handles_labels()
lines_2, labels_2 = ax2.get_legend_handles_labels()
ax1.legend(lines_1 + lines_2, labels_1 + labels_2, loc='upper left')

plt.tight_layout()
plt.show()
```

Типичный размер средств, собранных в рамках одного раунда, был максимальным в 2005 году.
В 2014 году наблюдается дальнейшее увеличение раундов финансирования по сравнению с 2013 и 2012 годами (14898 раундов против 10803 в 2013м и 6222 в 2012м), а также незначительный рост среднего объема финансирования (1млн 380 тыч долларов США).

```python
# Создаем сводную таблицу с размером финансирования по сегментам
pivot_market_year = pd.pivot_table(
  data=df,
  index='market',
  columns='year',
  values='funding_total_usd',
  aggfunc={'funding_total_usd':'sum'}
  )
# Отбираем сегменты, в которых сумма финансирования в 2014 была больше, чем в 2013
pivot_market_year = pivot_market_year[pivot_market_year[2014] > pivot_market_year[2013]]
# Выводим таблицу
pivot_market_year.head()
```

### 4.3 Годовая динамика доли возвращённых средств по типам финансирования

Поскольку заказчик хочет знать, какая часть вложенных или выданных денег со временем возвращается обратно инвесторам или финансистам, необходимо для каждого года и каждого вида финансирования рассчитать нормированные значения возврата средств: то есть какую долю возвращённые средства составляют от предоставленных. При этом слишком большие аномальные значения, то есть неадекватные выбросы, нужно заменить на пропуски.

Построим график, на котором отобразим нормированные значения возврата средств для типов финансирования `venture`, `debt_financing`, `private_equity`, `seed` и `angel`.

```python
# Создаем новый датафрейм, сгруппировав типы финансирования по годам и заменив нули на 1е-60
df1 = df.groupby('year')[funding_types].sum().replace(0.00, 1e-60)
# Создаем второй датафрейм с возвратами, приведя числа к одинаковому с первым датафреймом формату
df2 = cb_returns * 1000000
# Высчитываем процент возврата, заменив аномальные значения пропусками
df3 = ((df2 / df1).mask((df2 > 10 * df1) & (df1 != 0) & (df2 != 0), None)) * 100
```

```python
# Выводим необходимые для визуализации столбцы
df3[['venture','debt_financing','private_equity','seed','angel']]
```

```python
cols = ['venture', 'debt_financing', 'private_equity', 'seed', 'angel']

# Рисуем линейную диаграмму
df3.loc[2005:2014][cols].plot(
    kind='line', 
    figsize=(12, 6),
    marker='o'
)

# Настраиваем диаграмму
plt.title('Возвраты по типам финансирования', fontsize=14)
plt.xlabel('Год')
plt.ylabel('Значение в %')
plt.legend(title='Тип финансирования', loc='upper right')
plt.grid(axis='y')
plt.xticks(list(range(2005, 2015)), rotation=0)
plt.tight_layout()

plt.show()
```

Самым стабильным типом финансирования, судя по возвратам вложений, будет долговое финансирование. Стоит также обратить внимание на прямые/частные вложения, которые все еще показывают свою эффективность, несмотря на спад с 2009 года.

Была проведена объемная работа по предобработке, работе с выбросами и анализу данных, сопровождавшихся различными визуализациями. Были рассмотрены различные показатели предоставленных датасетов, выявлены доли сегментов на рынке, динамика привлеченных инвестиций и тенденции рынка в целом. Исходя из вышесказанного, сожно сделать следующие выводы:
- Больше всего на рынке компаний с единичным финансированием, однако по его объему лидируют компании с финансированием более года, большая часть которых относится к массовому сегменту;
- Венчурные инвестиции являются самыми популярными и по объему вложений, и по количеству компаний, и по возврату, однако самым эффективным в процентном соотношении типом остается долговое финансирование;
- С 2012 года наблюдается незначительный рост суммы вложений за каждый раунд финансирования, а количество самих раундов растет гораздно быстрее;
- Самыми быстрорастущими за последние несколько лет сегменты это программное обеспечение и биотехнологии. Однако остается немало других отраслей, показавших рост в 2014 по сравнению с 2013 годом.
